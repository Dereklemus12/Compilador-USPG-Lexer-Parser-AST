# Pruebas de integración

Se prueban entradas válidas, errores de sintaxis, precedencia, y la distinción entre errores semánticos y sintácticos.

```python
from pathlib import Path
import sys
raiz = Path.cwd()
if not (raiz / 'main.py').exists():
    raiz = raiz.parent
sys.path.insert(0, str(raiz.resolve()))
print('Raíz del proyecto:', raiz.resolve())
```

```python
from main import compilar
for archivo in sorted((raiz / 'pruebas').glob('*.txt')):
    contenido = archivo.read_text(encoding='utf-8')
    try:
        tokens, ast = compilar(contenido)
        print(archivo.name, '=> ACEPTADO, nodos:', len(ast.sentencias))
    except SyntaxError as exc:
        print(archivo.name, '=> ERROR:', exc)
```

**Conclusión:** los errores sintácticos se detectan; las referencias a identificadores no declarados se aceptan sintácticamente porque no se implementa análisis semántico.
