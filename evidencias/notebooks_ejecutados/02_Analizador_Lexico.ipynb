# Integración del lexer original

Se reutilizan los patrones y tipos del lexer anterior; se añade EOF y un error explícito ante caracteres desconocidos.

```python
from pathlib import Path
import sys
raiz = Path.cwd()
if not (raiz / 'main.py').exists():
    raiz = raiz.parent
sys.path.insert(0, str(raiz.resolve()))
print('Raíz del proyecto:', raiz.resolve())
```

```python
from lexer.lexer import Lexer
codigo = (raiz / 'pruebas/01_basico_valido.txt').read_text(encoding='utf-8')
tokens = Lexer(codigo).analizar()
for token in tokens: print(token)
assert tokens[-1].tipo == 'EOF' 
```
